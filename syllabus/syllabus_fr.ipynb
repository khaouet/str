# Systèmes Temps Réel et Ordonnancement Temps Réel
## Syllabus – 14 semaines

## 🎯 Objectifs du cours

- Comprendre les contraintes et la classification des systèmes temps réel.
- Maîtriser les modèles de tâches et les algorithmes d’ordonnancement monoprocesseur et multiprocesseur.
- Savoir gérer les ressources partagées et éviter les problèmes d’inversion de priorité.
- Être capable de concevoir et valider une application temps réel sur un RTOS.

## 📚 Prérequis

- Systèmes d’exploitation
- Programmation C
- Notions d’algorithmique et de structures de données

## 🗓️ Planning détaillé sur 14 semaines

| Semaine | Thème | Contenu principal | Activités |
|---------|-------|-------------------|-----------|
| **1** | Introduction aux systèmes temps réel | Définitions, exemples (avionique, automobile, robotique), contraintes temporelles, systèmes durs/mous/fermes, déterminisme, sûreté de fonctionnement | Cours + TP de prise en main d’un RTOS (ex. FreeRTOS) |
| **2** | Modélisation des tâches | Paramètres temporels : C, T, D, P, offset, jitter. Tâches périodiques, sporadiques, apériodiques. Contraintes de précédence et de ressources | Cours + TD sur hyperpériode et propriétés de base |
| **3** | Ordonnancement à priorité fixe : RM et DM | Rate Monotonic, Deadline Monotonic. Conditions d’ordonnançabilité (suffisante, nécessaire et suffisante). Analyse de temps de réponse (RTA) | Cours + TD/TP simulation |
| **4** | Ordonnancement dynamique : EDF et LLF | Earliest Deadline First, Least Laxity First. Optimalité d’EDF sur monoprocesseur. Comparaison RM/EDF, gestion de surcharge | Cours + TD comparaison d’algorithmes |
| **5** | Gestion des ressources partagées | Inversion de priorité, sections critiques. Protocoles : héritage de priorité (PIP), plafond de priorité (PCP, SRP), blocage | Cours + TP mise en évidence et résolution de l’inversion de priorité |
| **6** | Ordonnancement apériodique et serveurs | Tâches apériodiques/sporadiques. Serveurs : polling, deferrable server, sporadic server, priority exchange. Gestion des interruptions | Cours + TD/TP implémentation d’un serveur sporadique |
| **7** | **Évaluation théorique** | **Contrôle continu / partiel** couvrant les semaines 1 à 6 | Examen écrit |
| **8** | Ordonnancement multiprocesseur et distribué | Ordonnancement partitionné, global, semi-partitionné. Algorithmes : RM/EDF multiprocesseur, Pfair. Temps réel distribué : bus CAN, Ethernet temps réel | Cours + TD analyse simplifiée |
| **9** | Noyaux temps réel (RTOS) | Architecture d’un RTOS, ordonnanceur, interruptions, timers, files, mutex, latences, commutation de contexte. Exemples : FreeRTOS, Zephyr, VxWorks | Cours + TP développement multitâche |
| **10** | Analyse temporelle et validation | Pire temps d’exécution (WCET) : mesure, analyse statique. Traçabilité, tests, normes (DO-178C, ISO 26262) | Cours + TP mesure de WCET et analyse de traces |
| **11-12** | **Évaluation TP** | Projet pratique noté : conception et implantation d’un ordonnanceur temps réel (RM/EDF) ou application complète sur RTOS avec gestion de ressources et échéances | Évaluation pratique + rapport |
| **13** | Sujets avancés / Étude de cas | Systèmes critiques : avionique (ARINC 653), automobile (AUTOSAR/OSEK), systèmes cyber-physiques, tendances (temps réel et IA) | Cours / conférence / retour sur projet |
| **14** | **Révision générale** | Synthèse des concepts clés, annales, questions/réponses, préparation à l’examen final | Séance interactive |

## 📝 Modalités d’évaluation suggérées

- **Évaluation théorique (semaine 7)** : 40 %  
- **Évaluation TP (semaines 11-12)** : 50%+50%=100%  
- **Examen final éventuel ou contrôle continu** : 60%
